```python
# StyleDNA - Phase 2: Data Exploration
# Importing libraries

import pandas as pd
import matplotlib.pyplot as plt
```

```python
#Load the dataset
df = pd.read_csv('../data/raw/archive/styles.csv',on_bad_lines='skip')

# Preview the first 5 rows
df.head()
```

```python
# How much data do we have?
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("\nColumn names:")
print(df.columns.tolist())
```

```python
# Check for missing values
df.isnull().sum()
```

```python
# Drop rows with missing values in columns we care about
df_clean = df.dropna(subset=['baseColour', 'season', 'usage', 'productDisplayName'])

# Confirm how many rows we kept
print("Rows after cleaning:", df_clean.shape[0])
```

```python
# Save cleaned data
df_clean.to_csv('../data/cleaned/styles_cleaned.csv', index=False)

print("Cleaned data saved!")
```

```python
# Verify the cleaned file loads correctly
df_verify = pd.read_csv('../data/cleaned/styles_cleaned.csv')
print("Rows:", df_verify.shape[0])
print("Columns:", df_verify.shape[1])
print("Looks good!")
```
